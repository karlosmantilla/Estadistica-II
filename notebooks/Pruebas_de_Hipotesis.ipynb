<h2>¿Qué es una hipótesis?</h2>

Una hipótesis es una afirmación relativa a un parámetro de la población sujeta a verificación. El caso del peso corporal, nosostros hemos asumido que el peso pormedio es 72 kg, esta afirmación es una hipótesis y por consiguiente, debemos verificarla. Si tras la verificación encontramos que el peso promedio no es 72 kg, entraríamos a rechazar nuestra hipótesis y a tomar como verdadero el nuevo enunciado.

<h3>¿Cuáles son los pasos para contrastrar una hipótesis?</h3>

El procedimiento para contraste de hipótesis se puede resumir de la siguiente manera:

1. Se establecen las hipótesis (Nula y Alternativa)
2. Se selecciona el nivel de significancia
3. Se identifica el estadístico de prueba
4. Se formula la regla de decisión
5. Se toma una muestra
6. Se llega a una decisión

Veámoslos uno a uno:

<h2>Paso 1: Se plantean las hipótesis</h2>

Se requiren dos hipótesis, una que simbolice la afirmación y otra que simbolice el resultado alternativo. Por lo general, en se denominan en ese orden: _hipótesis nula_ ($H_{0}$) e _hipótesis alternativa_ ($H_{1}$).

* **_Hipótesis Nula:_** Es el enunciado relativo al valor de un parámetro poblacional que se formula con el fin de probar evidencia numérica.
* **_Hipótesis Alternativa:_** Es el **enunciado** que se acepta si los datos de la muestra ofrecen suficiente evidencia para rechazar la hipótesis nula.

Aunque es un aspecto que pareciera gramatical, ninguna hipótesis se acepta, la decisión gira en torno a la hipótesis nula y esta simplemente se rechaza o no se rechaza, esta forma de expresarlo tiene relación con el hecho de estar empleando una variable aleatoria.

**¿Como serían las hipótesis en el ejemplo del peso corporal?**

Como estamo afirmando que el peso promedio es 72 kg, esta afirmación es nuestra hipótesis nula. La negación, sería nuestra hipótesis alternativa:

$$
H_{0}: \mu = 72 \qquad H_{1}: \mu \neq 72
$$

**Para tener en cuenta:** $H_{0}$ siempre incluye el símbolo $=$ pues representa la información que se va a probar. Si la afirmación que hacemos se refiere a que un parámetro es menor o mayor que un valor, las hipótesis se modifican un poco en su escritura:

$$
H_{0}: \mu \leq 72 \qquad H_{1}: \mu > 72
$$

$$
H_{0}: \mu \geq 72 \qquad H_{1}: \mu < 72
$$

Siempre usamos la negación lógica en las hipótesis salvo en el caso de parámetros específicos, como la varianza, que tienen cotas en sus valores.

<h2>Paso 2: Se selecciona el nivel de significancia</h2>

El nivel de significancia representa la probabilidad de rechazar la hipótesis nula cuando es verdadera. De aquí se desprenden dos conceptos: _Error tipo I_ y _Error tipo II_:

* _Error tipo I_: Rechazar la hipótesis nula, $H_{0}$, cuando es verdadera ($\alpha$)
* _Error tipo II_: No Rechazar la hipótesis nula, $H_{0}$, cuando es falsa ($\beta$)

|$H_{0}$/Decisión| No Rechazar $H_{0}$ 	| Rechazar $H_{0}$ 	|
|:---:	|:---:	|:---:	|
| $H_{0}$ es verdadera 	| Decisión Correcta 	| Error Tipo I 	|
| $H_{0}$ es falsa 	| Error Tipo II 	| Decisión Correcta 	|

En nivel de significancia es el complemento del nivel de confianza, en general, vamos a encontrar miles de ejemplos que usan un nivel de $0.05$ (si no es que todos los ejemplos de los libros de texto los usan), pero el investigador puede usar el nivel de significancia que más se adapte a sus necesidades. Por simplicidad es que se recurre a este valor.

````r
levelconf <- 0.95
alpha2 <- (1-levelconf)/2
zeta <- qnorm(1-alpha2)
print(
paste0('El valor de Z para un nivel de significancia de ',2*alpha2,' es de :', round(zeta,2))
)
````

<h2>Paso 3: Seleccionar el estadístico de prueba</h2>

El estadístico de prueba corresponde al valor determinado a partir de la información de la muestra para establecer si se rechaza o no la hipótesis nula. Existen múltiples estadísticos pero los más usados son: $Z$, $t$, $F$ y $\chi^{2}$.

Además del tipo de variable, que definirá la distribución a emplear, se debe considerar la siguiente situación:

* Si se conoce $\sigma^{2}$ se emplea

$$
z = \frac{\overline{x} - \mu}{\sigma/\sqrt{n}}
$$

````r
library(RcmdrMisc)
options(repr.plot.width=7, repr.plot.height=7)
x <- seq(-3, 3, length.out=1000) 
plotDistr(x, dnorm(x, mean=0, sd=1), cdf=FALSE, xlab="x", ylab="Density", 
                   main=paste("Distribución Normal Estándar (Z)"), 
 regions=list(c(-1.96, 1.96)), col=c('#BEBEBE', '#BEBEBE'), legend.pos='topright')
````

* Si no se conoce $\sigma^{2}$ se usa

$$
t = \frac{\overline{x} - \mu}{s/\sqrt{n}}
$$

````r
# La Gráfica de la distribución t-student:
curve(dt(x, df = 30), from = -3, to = 3, lwd = 3, ylab = "y", main = "Distribución t-student")
ind <- c(1, 2, 3, 5, 10)
for (i in ind) curve(dt(x, df = i), -3, 3, add = TRUE,col=i)
````

<h2>paso 4: Se formula la regla de decisión</h2>

Para tomar la decisión sobre $H_{0}$ es necesario establecer un valor de referencia para el estadístico de prueba. Para ello se requiere un valor crítico que corresponde al punto de división entre la región donde se rechaza la hipótesis nula y la región donde no se rechaza. En la gráfica de $Z$ estamos señalando en blanco la región de No Rechazo y en gris la región de rechazo:

````r
library(dplyr)
df <- data.frame(x = seq(-3, 3, length.out = 100)) %>% mutate(y = dnorm(x))

library(ggplot2)
library(gghighlight)

z1 <- ggplot(df, aes(x, y)) +
geom_area(fill = "white", color = 'black') +
gghighlight(x < 1.96 & x > -1.96) +
ggtitle(expression(paste('H1 : ', bar(x), ' ≠ ', mu))) +
theme(axis.text = element_text(size = 10),
     axis.title = element_text(size = 10),
     plot.title = element_text(size=12, hjust = 0.5))

z2 <- ggplot(df, aes(x, y)) +
geom_area(fill = "white", color = 'black') +
gghighlight(x < 1.96) +
ggtitle(expression(paste('H1 : ', bar(x), ' > ', mu))) +
theme(axis.text = element_text(size = 10),
     axis.title = element_text(size = 10),
     plot.title = element_text(size=12, hjust = 0.5))

z3 <- ggplot(df, aes(x, y)) +
geom_area(fill = "white", color = 'black') +
gghighlight(x > -1.96) +
ggtitle(expression(paste('H1 : ', bar(x), ' < ', mu))) +
theme(axis.text = element_text(size = 10),
     axis.title = element_text(size = 10),
     plot.title = element_text(size=12, hjust = 0.5))

options(repr.plot.width=14, repr.plot.height=6)
ggarrange(z1,z2,z3, ncol = 3, nrow = 1)
````

Tenemos un concepto adicional: _valor-p_ que representa la probabilidad de observar un valor muestral tan extremo o más que el valor observado si la hipótesis nula es verdadera. Con base en esto, tenemos los siguientes criterios de decisión:

Si tomamos el valor de $Z$:

* Rechazar $H_{0}$ si $Z > Z_{\alpha}$
* No Rechazar $H_{0}$ si $Z < Z_{\alpha}$


Si tomamos el valor de $\alpha$:

* Rechazar $H_{0}$ si $p-valor << \alpha$
* No Rechazar $H_{0}$ si $p-valor > \alpha$

<h2>Tipos de Pruebas</h2>

En pruebas de hipótesis podemos contar con una o más muestras, vamos a abordar el caso en el que contamos con una o dos muestras que queremos comparar.

<h3>Pruebas de hipótesis para una muestra</h3>

Para el casi de una sola muestra tenemos las siguientes opciones:

* **Prueba para la media si se conoce la varianza:**

$$
z = \frac{\overline{x} - \mu}{\sigma/\sqrt{n}}
$$


* **Prueba para la media si NO se conoce la varianza:**

\begin{equation}
t = \frac{\overline{x} - \mu}{s/\sqrt{n}}
\end{equation}


* **Para la proporción ($p$):**

\begin{equation}
Z = \frac{p - \pi}{\sqrt{\frac{\pi (1 - \pi)}{n}}}
\end{equation}

<h3>Pruebas de hipótesis para dos muestras independientes</h3>

* **Para dos medias:**
\begin{equation}
Z = \frac{\overline{x}_{1} - \overline{x}_{2}}{\sigma_{\overline{x}_{1} - \overline{x}_{2}}}
\end{equation}

donde $\sigma_{\overline{x}_{1} - \overline{x}_{2}}^{2}$ es la varianza de la distribución de la diferencia

$$
\sigma_{\overline{x}_{1} - \overline{x}_{2}}^{2} = \frac{\sigma_{1}^{2}}{n_{1}} + \frac{\sigma_{2}^{2}}{n_{2}}
$$


* **Para Proporciones:**

\begin{equation}
Z = \frac{p_{1} - p_{2}}{\sqrt{\frac{p_{c}(1-p_{c})}{n_{1}} + \frac{p_{c}(1-p_{c})}{n_{2}}}}
\end{equation}

donde $p_{c}$ es la proporción conjunta

$$
p_{c} = \frac{x_{1} - x_{2}}{n_{1} + n_{2}}
$$

* **Para Varianzas Iguales Desconocidas:**

\begin{equation}
t = \frac{\overline{x}_{1} - \overline{x}_{2}}{\sqrt{s_{p}^{2} \left( \frac{1}{n_{1}} + \frac{1}{n_{2}} \right) }}
\end{equation}

donde $s_{p}^{2}$ es la varianza conjunta y se representa:

$$
s_{p}^{2} = \frac{ (n_{1} - 1)s_{1}^{2} + (n_{2} - 1)s_{2}^{2} }{n_{1} + n_{2} - 2}
$$

* **para Varianzas Desiguales Desconocidas: **

Como las varianzas son desiguales no existe varianza conjunta.

\begin{equation}
t = \frac{\overline{x}_{1} - \overline{x}_{2}}{\sqrt{\frac{s_{1}^{2}}{n_{1}} + \frac{s_{2}^{2}}{n_{2}}}}
\end{equation}

Para calcular los grados de libertad de la prueba:

\begin{equation}
gl = \frac{\left[ \frac{s_{1}^{2}}{n_{1}} + \frac{s_{2}^{2}}{n_{2}} \right]^{2}}{\frac{\left( \frac{s_{1}^{2}}{n_{1}} \right)^{2}}{n_{1}-1} + \frac{\left( \frac{s_{2}^{2}}{n_{2}} \right)^{2}}{n_{2}-1}}
\end{equation}

* **Para Muestras Dependientes:**

\begin{equation}
t = \frac{\overline{d}}{\frac{s_{d}}{\sqrt{n}}}
\end{equation}

donde:

$$
s_{d} = \sqrt{ \frac{\sum \left( d - \overline{d} \right)^{2}}{n - 1} }
$$

## Pruebas para proporciones

### Una proporción

Hemos venido trabajando con un conjunto de datos que representa las ventas de una empresa estadounidense. A partir de lo que ya exploramos, surge una pregunta interesante:
¿será cierto que más de la mitad de las ventas corresponden a Estados Unidos?

Podemos traducir esta inquietud en una prueba de hipótesis:

$$
H_{0} : p_{USA} \leq 0.5 \qquad H_{1}: p_{USA} > 0.5
$$

donde $p_{USA}$ representa la proporción de ventas realizadas en Estados Unidos.

Para discutir en clase:
¿Cómo podríamos abordar esta prueba de hipótesis con base en los datos que ya conocemos?
¿Qué condiciones deberíamos verificar antes de aplicar la prueba? ¿Y cómo interpretaríamos el resultado si, por ejemplo, obtenemos un valor-p de 0.03?

Al organizar la información en una tabla de frecuencias, podemos ver con mayor claridad qué está ocurriendo con los datos:

````r
##  Tabla de Frecuencias
with(m.datos, table(Country))
````

Ahora enfoquémonos en lo que ocurre desde la perspectiva de las proporciones:

````r
with(m.datos, prop.table(table(Country)))
````

Se trata de una situación simple: verificar si la proporción de registros de USA es mayor a 0.5. A continuación, formularemos la prueba de hipótesis:

Primero, revisemos la función que vamos a utilizar para aplicar la prueba de proporciones.

````r
# Veamos la función que vamos a usar:
args(prop.test)
````

Vamos a realizar la prueba de hipótesis con base en el siguiente planteamiento:

$$
H_{0} : p_{USA} \leq 0.5 \qquad H_{1}: p_{USA} >0.5
$$

Este planteamiento nos servirá como guía para entender mejor cómo introducir los argumentos en la función. Procedamos con la aplicación de la prueba.

````r
conf.lev <- 0.95
sig.lev <- 1 - conf.lev
p.prop <- 0.5
# Realizamos la prueba de hipótesis:
p.prob <- prop.test(nrow(m.datos[m.datos$Country == 'USA',]), 
          nrow(m.datos),
          p = p.prop,
          alternative = "greater", conf.level = conf.lev, correct = TRUE)

p.prob
````

````r
str(p.prob)
````

````r
library(dplyr)
````

````r
if_else(p.prob$p.value < sig.lev, 'Decisión: Rechazar H0', 'Decisión: No Rechazar H0')
````

##### Conclusión:

Veamos cómo podemos concluir:

````r
print(paste0('Con un nivel de significancia de ',
             sig.lev,
             ' la evidencia sugiere que la proporción de registros de ventas para USA',
             if_else(p.prob$p.value < sig.lev, ' es mayor que ', ' no es mayor que '), p.prop))
````

### Una media con varianza conocida

Dado el volumen observado en las ventas, sospechamos que el promedio diario de ventas en Estados Unidos supera las 410 unidades monetarias.

Planteemos la situación de forma estructurada:

- Hipótesis:

$$
H_{0}: \bar{X}_{USA} \leq 410 \qquad H_{1}: \bar{X}_{USA} > 410

````r
m.usa <- m.datos[m.datos$Country == 'USA','Revenue']
conf.lev <- 0.95
sig.lev <- 1 - conf.lev
m.prueba <- 410

p.zeta <- z.test(m.usa,
       sd = sd(m.usa),
       mu = m.prueba,
       alternative = 'greater',
      conf.level = conf.lev)

p.zeta
````

# Métodos No Paramétricos en Estadística

Muchas de las pruebas utilizadas en estadística se basan en el supuesto de que los datos provienen de una distribución normal. Sin embargo, existen situaciones en las que no es necesario asumir normalidad en la población.

Para estos casos, contamos con una alternativa: las **pruebas no paramétricas**, que permiten realizar inferencias sin requerir supuestos estrictos sobre la forma de la distribución.

## Prueba de Bondad de Ajuste

La prueba de bondad de ajuste es una de las herramientas estadísticas más utilizadas, especialmente cuando se trabaja con variables categóricas. Su utilidad radica en que no requiere suponer una escala de medición numérica, sino que puede aplicarse incluso con datos en escala nominal, es decir, datos clasificados en categorías o grupos.

El objetivo principal de esta prueba es comparar una distribución observada con una distribución teórica o esperada, para evaluar si existen diferencias significativas entre ambas.

El estadístico utilizado es *chi-cuadrada* ($\chi^{2}$):

$$
\chi^{2} = \sum \left[ \frac{(f_{o} - f_{e})^{2}}{f_{e}} \right]
$$

Este estadístico se distribuye según una chi-cuadrado con $k-1$ grados de libertad, donde:

* $k$ es el número de categorías
* $f_{o}$ es la frecuencia observada en una categoría específica,
* $f_{e}$ es la frecuencia esperada para esa categoría.

Las hipótesis se plantean:

* $H_{0}:$ No hay diferencia entre las proporciones de las categorías.
* $H_{1}:$ Existe diferencia entre las proporciones de las categorías.## Prueba de Bondad de Ajuste

````r
x <- rchisq(10000, 5)
hist(x, prob=TRUE)
curve( dchisq(x, df=5), col='green', lwd = 2, add=TRUE)
curve( dchisq(x, df=10), col='red', lwd = 2, add=TRUE )
curve( dchisq(x, df=15), col='blue', lwd = 2, add=TRUE )
curve( dchisq(x, df=30), col='black', lwd = 2, add=TRUE )
````

## Prueba de Independencia (Tablas de Contingencia)

El estadístico $\chi^{2}$ también se utiliza para evaluar si existe una relación entre dos variables categóricas, es decir, para comprobar si una variable es independiente de la otra.

Este tipo de análisis se realiza a partir de una tabla de contingencia, que resume las frecuencias observadas para todas las combinaciones posibles entre las categorías de ambas variables.

La fórmula del estadístico es:

$$
\chi^{2} = \sum_{i=1}^{r} \sum_{j=1}^{k} \frac{ \left( n_{ij} - e_{ij} \right)^{2} }{e_{ij}}
$$

donde:
- $n_{ij}$ es la frecuencia observada en la celda correspondiente a la fila $i$ y columna $j$,
- $e_{ij}$ es la frecuencia esperada en esa misma celda,
- $r$ es el número de filas,
- $k$ es el número de columnas.

Los **grados de libertad** se calculan como:

$$
gl = (r - 1)(k - 1)
$$

La **frecuencia esperada** para cada celda se obtiene aplicando la siguiente fórmula:

$$
e_{ij} = \frac{n_{i \cdot} \times n_{\cdot j}}{n}
$$

donde:

- $n_{i \cdot}$ es el total de la fila $i$,
- $n_{\cdot j}$ es el total de la columna $j$,
- $n$ es el total general (suma de todas las frecuencias observadas).

### Veamos un ejemplo

Supongamos que queremos determinar si la **preferencia por el sabor del helado** está relacionada con el **sexo de las personas**. En otras palabras, queremos saber si **la elección del sabor es independiente del sexo**.

Planteamos las siguientes hipótesis:

- $H_0$: La preferencia del sabor del helado es independiente del sexo.  
- $H_1$: La preferencia del sabor del helado depende del sexo.

La información se resume en la siguiente tabla de contingencia:

| Sexo     | Chocolate | Vainilla | Fresa |
|----------|:---------:|:--------:|:-----:|
| Hombres  |    100    |   120    |  60   |
| Mujeres  |    350    |   200    |  90   |

### ¿Cómo lo resolvemos en R?

A continuación, aplicaremos la prueba de independencia utilizando esta tabla.


````r
# Asignamos las frecuencias:

Hombres = c(100, 120, 60)
Mujeres = c(350, 200, 90)

# Construimos el data frame:
helados = as.data.frame(rbind(Hombres, Mujeres))

# Asignamos nombres a las variables
names(helados) = c('Chocolate', 'Vainilla', 'Fresa')

# La tabla queda:
helados
````

````r
# Ahora, la prueba:

chisq.test(helados)
````

## Prueba de los rangos con signo de Wilcoxon (datos ordenados)

La **prueba de los rangos con signo de Wilcoxon**, también conocida como *Wilcoxon signed-rank test*, es un test **no paramétrico** que permite comparar dos muestras relacionadas cuando no se cumplen los supuestos necesarios para aplicar pruebas paramétricas.

Esta prueba es una **alternativa a la prueba t para muestras dependientes (pareadas)**, especialmente útil cuando:

- Las distribuciones son asimétricas,
- Existen valores atípicos,
- O el tamaño de la muestra es tan pequeño que no se puede verificar con seguridad si provienen de poblaciones normales.

### ¿Cuándo utilizar esta prueba?

Se recomienda aplicar el *Wilcoxon signed-rank test* en los siguientes casos:

- Cuando los **datos son dependientes** (por ejemplo, mediciones antes y después sobre los mismos individuos).
- Cuando los datos se pueden **ordenar** (es decir, son al menos de tipo ordinal).
- **No se requiere normalidad** en la distribución de las diferencias, pero sí se asume que estas diferencias son **simétricas**.
- A diferencia del t-test, esta prueba se basa en las **medianas** y no en las medias.
- Es especialmente recomendable cuando hay **valores extremos**, **falta de normalidad** o **tamaños de muestra pequeños**.

El **estadístico de prueba** en la prueba de los rangos con signo de Wilcoxon es sencillo:  
se calcula como el **mínimo entre la suma de los rangos positivos y la suma de los rangos negativos**:

$$
W = \min(W^+, W^-)
$$

Donde:

- $W^+$ es la suma de los rangos con signo positivo.  
- $W^-$ es la suma de los rangos con signo negativo.

### Para tener en cuenta:

- Si el tamaño de la muestra es **menor a 25** ($n < 25$), se utiliza la **tabla de Wilcoxon** para determinar la significancia.
- Si el tamaño es **mayor o igual a 25** ($n \geq 25$), se puede aproximar la distribución mediante la **distribución normal estándar**, usando la siguiente fórmula:

$$
Z = \frac{W - \frac{n_1 n_2}{2}}{\sqrt{ \frac{n_1 n_2 (n_1 + n_2 + 1)}{12} }} \sim N(0, 1)
$$

### Ejemplo

Supongamos que contamos con **dos muestras pareadas**, pero **no conocemos la distribución de las poblaciones** de las que provienen. Además, el **tamaño de la muestra es pequeño**, lo que dificulta verificar si siguen una distribución normal.

¿Existe una diferencia significativa entre ambas condiciones?

> **Nota:** En este ejemplo se utilizan muestras pequeñas con fines ilustrativos, para mostrar con claridad los pasos del procedimiento. Sin embargo, debe tenerse en cuenta que el *Wilcoxon signed-rank test* puede no ser muy preciso con tamaños de muestra extremadamente reducidos.

### Hipótesis

- $H_0$: La mediana de las diferencias es igual a cero (no hay cambio).  
- $H_1$: La mediana de las diferencias es distinta de cero (hay un cambio significativo).

````r
# Escribamos los resultados:

antes <- c( 2, 5, 4, 6, 1, 3 )
despues <- c( 5, 6, 2, 7, 1, 6 )
````

````r
# Calculamos las diferencias
diferencias <- c(antes - despues)
rbind(antes, despues, diferencias)
````

````r
# Se asigna a cada valor absoluto de las diferencias un rango (posición), comenzando con el valor 1 para la diferencia
# más pequeña.
# Si hay valores repetidos (situación conocida como ligaduras o ties), se les asigna el promedio de las posiciones que 
# ocuparían.
# Las diferencias iguales a cero se excluyen del proceso de asignación de rangos.

rangosdiferencias <- rank(abs(diferencias[diferencias != 0])) 
# La función rank() de R calcula las posiciones automáticamente, solucionando las ligaduras en caso de que las haya.

rangosdiferencias <- c(rangosdiferencias[1:4], 0, rangosdiferencias[5]) 
# Se incorpora un 0 en la posición que ocupan las diferencias nulas para cuadrar las posiciones.

rangosdiferencias
````

````r
# Se organiza la información

tabla <- data.frame(antes = antes, despues = despues, signo = sign(diferencias),
                    diferencia = abs(diferencias), rangos = rangosdiferencias)
tabla
````

````r
# Se calcula el estadístico
sumaPositivos <- sum(tabla[tabla$signo == 1,"rangos"])
sumaPositivos
````

````r
sumaNegativos <- sum(tabla[ tabla$signo == -1,"rangos"] )
sumaNegativos
````

````r
W <- min(c( sumaPositivos, sumaNegativos ))
W
````

A continuación, debemos consultar la tabla de valores críticos del test de Wilcoxon para determinar si el estadístico calculado es significativo.

Podemos acceder a la tabla en el siguiente enlace: https://www.real-statistics.com/statistics-tables/wilcoxon-signed-ranks-table/

Veamos la función en el programa:

````r
wilcox.test(x = antes, y = despues, alternative = "two.sided", mu = 0, paired = TRUE)
````

En la salida que devuelve la función `wilcox.test()`, el **estadístico de prueba aparece identificado como `V`**, aunque corresponde conceptualmente al **estadístico \( W \)** del test de Wilcoxon.

Cuando existen **ligaduras** (*ties*) en los datos, `wilcox.test()` **no puede calcular el valor-p exacto**, por lo que retorna una **aproximación** basada en la suposición de que \( W \) sigue una distribución normal.

En estos casos —y también cuando los tamaños muestrales son mayores a 25— es recomendable utilizar la función `wilcoxsign_test()` del paquete `{coin}`, ya que esta sí permite obtener el **valor-p exacto**, sin necesidad de asumir normalidad.
